## Manipular arquivos csv

### Subtask:
Gerar código para ler e escrever em arquivos CSV usando a biblioteca `csv`.


**Reasoning**:
Generate code to write and read CSV files using the `csv` library as per the instructions.



```python
import csv

# Data to be written to the CSV file
data = [
    ['Name', 'Age', 'City'],
    ['Alice', 30, 'New York'],
    ['Bob', 25, 'London'],
    ['Charlie', 35, 'Paris']
]

csv_file_name = "example.csv"

# Write data to the CSV file
with open(csv_file_name, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)

print(f"Data written to {csv_file_name}")

# Read data from the CSV file
print(f"\nReading data from {csv_file_name}:")
with open(csv_file_name, mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Manipular arquivos json

### Subtask:
Gerar código para ler e escrever em arquivos JSON usando a biblioteca `json`.


**Reasoning**:
Import the `json` library, define data to be written to a JSON file, specify the file name, write the data to the file with indentation, print a confirmation message, read the data back from the file, and print the read data.



```python
import json

# Data to be written to the JSON file
data = {
    "name": "John Doe",
    "age": 30,
    "isStudent": False,
    "courses": [
        {"title": "Math", "credits": 3},
        {"title": "Science", "credits": 4}
    ]
}

json_file_name = "example.json"

# Write data to the JSON file
with open(json_file_name, 'w') as f:
    json.dump(data, f, indent=4)

print(f"Data written to {json_file_name}")

# Read data from the JSON file
print(f"\nReading data from {json_file_name}:")
with open(json_file_name, 'r') as f:
    read_data = json.load(f)

print(read_data)
```

## Criar um diretório

### Subtask:
Gerar código para criar um novo diretório.


**Reasoning**:
The subtask is to create a new directory. I will import the `os` module, define a directory name, check if it exists, and then create it using `os.makedirs()`.

## Remover um diretório

## Obter informações sobre um arquivo

### Subtask:
Gerar código para obter informações como tamanho, data de modificação, etc., de um arquivo.


**Reasoning**:
The subtask requires getting file information, which involves importing the `os` and `datetime` modules, checking for file existence, using `os.stat()` to get file information, and then printing the size and modification time. These steps can be combined into a single code block.



```python
import os
import datetime

file_name = "renamed_example.csv"

if os.path.exists(file_name):
    file_info = os.stat(file_name)

    print(f"File Size: {file_info.st_size} bytes")

    # Convert modification time to a human-readable format
    mod_time_timestamp = file_info.st_mtime
    mod_time_datetime = datetime.datetime.fromtimestamp(mod_time_timestamp)
    print(f"Last Modification Time: {mod_time_datetime}")
else:
    print(f"The file '{file_name}' was not found.")
```
